```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Analisis: List Python menggunakan operator * untuk pengulangan (duplication) yang membutuhkan bilangan bulat. NumPy Array mendukung operasi berbasis vektor (element-wise), sehingga operasi matematika secara otomatis diterapkan ke setiap elemen di dalamnya.

```python
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

df = pd.DataFrame(data_kantin)
print(df)
```

Perhitungan Statistik Bias: Fungsi agregasi (seperti rata-rata atau total) bisa memberikan hasil yang keliru atau mengabaikan item penting.

Pengelompokan Tidak Valid: Pengelompokan data (groupby) berdasarkan kolom menu akan menciptakan kelompok kategori None yang tidak bermakna secara bisnis.

Sistem Error: Fungsi lanjutan atau model analisis berbasis data dapat mengalami kegagalan proses (error) ketika menerima input data kosong.

```python
# Membuat file data_kantin.csv sementara di Google Colab
df.to_csv('data_kantin.csv', index=False)

# Membaca data dari file CSV
df = pd.read_csv('data_kantin.csv')

print("--- 5 Baris Pertama ---")
print(df.head())

print("\n--- Info DataFrame ---")
print(df.info())

print("\n--- Statistik Ringkas ---")
print(df.describe())

print("\n--- Ukuran Dataset (Baris, Kolom) ---")
print(df.shape)
```

Makna Hasil Tersebut:
Terdapat 1 baris data kosong (missing value) pada kolom menu dan 1 baris data kosong pada kolom terjual. Hal ini menandakan bahwa dataset belum utuh dan memerlukan penanganan (imputation atau deletion) sebelum analisis dila

```python
print("--- Jumlah Data Kosong Sebelum ---")
print(df.isnull().sum())

# Penanganan missing values
df['terjual'] = df['terjual'].fillna(0)  # Isi kekosongan kolom terjual dengan 0
df = df.dropna(subset=['menu'])         # Hapus baris jika kolom menu kosong

print("\n--- Jumlah Data Kosong Sesudah ---")
print(df.isnull().sum())
```

Mengapa terjual Diisi (fillna(0)):

Logika Bisnis: Jika nama menu tercantum tetapi angka penjualannya kosong, asumsi paling aman adalah menu tersebut tidak terjual sama sekali (0 unit) pada periode analisis.

Menjaga Informasi Produk: Mengisi angka 0 mempertahankan baris data tersebut sehingga variabel harga dan nama menu tetap bisa dipakai untuk analisis harga.

Mengapa menu Dihapus (dropna(subset=['menu'])):

Kehilangan Identitas: Data penjualan dan harga tanpa nama produk tidak memiliki nilai konteks bisnis. Kita tidak tahu produk apa yang menghasilkan angka tersebut, sehingga menghapus baris adalah opsi terbaik agar tidak merusak validitas analisis produk.

```python
print('Jumlah baris duplikat:', df.duplicated().sum())

# Menghapus duplikat
df = df.drop_duplicates()

# Memastikan tipe data integer
df['terjual'] = df['terjual'].astype(int)
df['harga'] = df['harga'].astype(int)

print("\n--- Tipe Data Kolom ---")
print(df.dtypes)
print("\n--- Data Setelah Duplikat Dihapus ---")
print(df)
```

Jumlah Baris Setelah drop_duplicates():
Jumlah baris berkurang jika terdapat baris yang seluruh nilainya identik pada semua kolom. Jika tidak ada baris yang 100% sama persis, jumlah baris tetap sama. (Catatan: Baris "Es Teh" pada dataset tidak dihapus karena memiliki nilai terjual yang berbeda, yaitu 40 dan 35).

Pentingnya Memastikan Tipe Data (dtypes) Benar:

Keakuratan Operasi Matematika: Memastikan kolom numerik (seperti harga) berjenis integer/float sangat vital. Jika harga bertipe string (teks), operasi harga * 2 akan melakukan penggabungan teks ('40004000') alih-alih perkalian angka (8000).

Efisiensi Memori: Tipe data yang tepat mengoptimalkan penggunaan memori dan kecepatan pemrosesan data.

```python
# 1. Filtering (terjual > 20)
laris = df[df['terjual'] > 20]
print("--- Menu Laris (>20) ---")
print(laris)

# 2. Sorting
urut = df.sort_values(by='terjual', ascending=False)

# 3. Kolom Turunan (total_pendapatan)
df['total_pendapatan'] = df['harga'] * df['terjual']

# 4. Agregasi Groupby
ringkasan = df.groupby('menu')['total_pendapatan'].sum()
print("\n--- Ringkasan Total Pendapatan per Menu ---")
print(ringkasan)
```

Menu Hasil total_pendapatan Tertinggi:Es Teh dengan total pendapatan sebesar Rp 300.000 (gabungan dari $40 \times 4000 = 160.000$ dan $35 \times 4000 = 140.000$), diikuti oleh Nasi Goreng sebesar Rp 276.000.Pemanfaatan Informasi untuk Pengambilan Keputusan Kantin:Manajemen Stok & Distribusi: Es Teh adalah kontributor pendapatan utama dan paling diminati. Pengelola kantin harus memastikan persediaan bahan baku (es, teh, gula) selalu cukup agar tidak kehabisan stok (stockout).Strategi Promosi/Bundling: Es Teh dapat digunakan sebagai produk pemancing (lead product) dalam paket hemat bersama menu yang kurang laris untuk menggenjot penjualan menu lain.Evaluasi Performa Menu: Menu dengan pendapatan 0 atau rendah (seperti Mie Ayam akibat data penjualan kosong/diisi 0) perlu dievaluasi dari sisi promosi, rasa, atau penyesuaian harga.

```python
import pandas as pd

# 1. Menyiapkan DataFrame (hasil dari Latihan 1 - 5)
data_kantin = {
    'menu': ['Nasi Goreng', 'Mie Ayam', 'Es Teh'],
    'harga': [12000, 10000, 4000],
    'terjual': [23, 0, 40]
}
df = pd.DataFrame(data_kantin)

# 2. Filtering & Sorting
laris = df[df['terjual'] > 20]
urut = df.sort_values(by='terjual', ascending=False)

# 3. Membuat kolom turunan total_pendapatan
df['total_pendapatan'] = df['harga'] * df['terjual']

# 4. Agregasi dengan Groupby
ringkasan = df.groupby('menu')['total_pendapatan'].sum()
print("--- Ringkasan Total Pendapatan per Menu ---")
print(ringkasan)

# 5. KODE TAMBAHAN: Mencari menu dengan total_pendapatan tertinggi secara otomatis
menu_tertinggi = ringkasan.idxmax()
pendapatan_tertinggi = ringkasan.max()

print("\n--- KESIMPULAN HASIL ANALISIS ---")
print(f"Menu dengan pendapatan tertinggi: {menu_tertinggi}")
print(f"Total pendapatan: Rp{pendapatan_tertinggi:,}")
```
