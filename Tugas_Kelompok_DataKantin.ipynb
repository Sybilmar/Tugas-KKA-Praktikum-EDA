```python
import pandas as pd

# Menyiapkan data awal
data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

# Menyimpan menjadi file CSV
df_awal = pd.DataFrame(data_kantin)
df_awal.to_csv('data_kantin.csv', index=False)
print("File 'data_kantin.csv' berhasil dibuat!")
```

Pesan Konfirmasi: Konsol menampilkan teks "File 'data_kantin.csv' berhasil dibuat!" sebagai penanda bahwa proses eksekusi kode berjalan lancar sampai akhir.

Pembuatan File CSV: Terbentuk file baru bernama data_kantin.csv di direktori kerja tanpa menyimpan nomor indeks baris (karena index=False).

Konversi Tipe Data Terjual: Nilai None pada baris ke-3 kolom 'terjual' otomatis diubah Pandas menjadi NaN (Not a Number), yang menyebabkan seluruh tipe data pada kolom 'terjual' dikonversi menjadi float (desimal, seperti 23.0, 40.0, dst.).

```python
import pandas as pd

# 1. Memuat dataset
df = pd.read_csv('data_kantin.csv')

# 2. Menampilkan 5 baris pertama
print("--- 5 Baris Pertama (head) ---")
print(df.head())

# 3. Menampilkan info kolom dan tipe data
print("\n--- Informasi Data (info) ---")
print(df.info())

# 4. Menampilkan statistik ringkas
print("\n--- Statistik Ringkas (describe) ---")
print(df.describe())

# 5. Menampilkan ukuran dataset
print("\n--- Ukuran Dataset (shape) ---")
print(f"Jumlah Baris: {df.shape[0]}, Jumlah Kolom: {df.shape[1]}")
```

df.head(): Menampilkan 5 baris pertama data kantin. Terlihat baris ke-4 memuat nilai NaN di kolom menu dan baris ke-2 memuat NaN di kolom terjual.

df.info(): Rincian struktur data menunjukkan total 5 baris data (index 0–4) dengan 3 kolom. Terdeteksi adanya missing value (non-null count berjumlah 4 dari 5) pada kolom menu dan terjual. Kolom menu bertipe object (teks), sedangkan harga bertipe int64 dan terjual bertipe float64.

df.describe(): Menampilkan ringkasan statistik khusus kolom numerik (harga dan terjual), seperti rata-rata (mean), nilai terkecil/terbesar (min/max), serta standar deviasi (std). Nilai count pada terjual bernilai 4 (bukan 5) karena data NaN diabaikan dalam perhitungan statistik.

df.shape: Mengembalikan nilai Jumlah Baris: 5, Jumlah Kolom: 3 yang menandakan ukuran keseluruhan matriks DataFrame.

```python
# 1. Cek jumlah missing values awal
print("--- Missing Values Awal ---")
print(df.isnull().sum())

# 2. Penanganan missing value (fillna & dropna)
df['terjual'] = df['terjual'].fillna(0)  # Isi terjual kosong dengan 0
df = df.dropna(subset=['menu'])         # Hapus baris jika menu kosong

# 3. Penanganan data duplikat
jumlah_duplikat = df.duplicated().sum()
print(f"\nJumlah duplikat ditemukan: {jumlah_duplikat}")
df = df.drop_duplicates()

# 4. Penyesuaian tipe data ke integer
df['harga'] = df['harga'].astype(int)
df['terjual'] = df['terjual'].astype(int)

print("\n--- Tipe Data Setelah Dibersihkan ---")
print(df.dtypes)

print("\n--- Data Setelah Pembersihan ---")
print(df)
```

Missing Values: Menghitung data kosong di awal (ditemukan 1 pada menu dan 1 pada terjual), lalu mengisi terjual yang kosong dengan 0 dan menghapus baris yang menu-nya kosong.

Duplikat: Mendeteksi 1 baris duplikat ('Es Teh') dan menghapusnya dari data.

Konversi Tipe Data: Mengubah tipe data kolom harga dan terjual menjadi int64 (bilangan bulat tanpa desimal).

Hasil Akhir: Dataset menjadi bersih tanpa data kosong maupun duplikat, terdiri dari 3 baris data dengan tipe data angka yang sudah presisi.

```python
# 1. Membuat kolom turunan total_pendapatan
df['total_pendapatan'] = df['harga'] * df['terjual']

# 2. Filtering (menu dengan terjual > 10)
df_laris = df[df['terjual'] > 10]
print("--- Data Produk Laris (Terjual > 10) ---")
print(df_laris)

# 3. Sorting (diurutkan berdasarkan total_pendapatan terbesar)
df_sorted = df.sort_values(by='total_pendapatan', ascending=False)
print("\n--- Data Diurutkan Pendapatan Tertinggi ---")
print(df_sorted)

# 4. Groupby & Agregasi per Menu
ringkasan_menu = df.groupby('menu').agg({
    'terjual': 'sum',
    'total_pendapatan': 'sum'
}).reset_index()

print("\n--- Ringkasan Total Penjualan per Menu ---")
print(ringkasan_menu)
```


Kolom Baru: Menambahkan kolom total_pendapatan hasil perkalian antara harga dan terjual.

Filtering: Menyaring dan menampilkan baris data yang nilai terjual-nya lebih besar dari 10.

Sorting: Mengurutkan data berdasarkan total_pendapatan dari nilai tertinggi ke terendah (ascending=False).

Groupby & Agregasi: Mengelompokkan data berdasarkan menu lalu menjumlahkan total unit terjual dan total_pendapatan untuk setiap menu secara terpisah.

```python
# 1. Simpan dataset bersih ke CSV
df.to_csv('dataset_bersih.csv', index=False)
print("[INFO] Dataset bersih berhasil disimpan ke 'dataset_bersih.csv'\n")

# 2. Mengambil data untuk ringkasan narasi
menu_teratas = df_sorted.iloc[0]['menu']
pendapatan_teratas = df_sorted.iloc[0]['total_pendapatan']
total_omset = df['total_pendapatan'].sum()

# 3. Menampilkan Data Profiling Summary
print("="*50)
print("DATA PROFILING SUMMARY (RINGKASAN NARASI)")
print("="*50)
print(f"1. Total omset penjualan kantin dari data yang dibersihkan adalah sebesar Rp{total_omset:,}.")
print(f"2. Produk dengan pendapatan tertinggi adalah '{menu_teratas}' dengan total pendapatan Rp{pendapatan_teratas:,}.")
print(f"3. Seluruh missing value dan duplikat telah berhasil ditangani, sehingga file 'dataset_bersih.csv' siap digunakan untuk tahap visualisasi data.")
```

Eksport File: Menyimpan DataFrame yang sudah rapi ke file baru bernama dataset_bersih.csv tanpa menyimpan nomor indeks.

Ekstraksi Informasi: Mengambil baris pertama dari data yang sudah diurutkan (df_sorted) untuk mendapatkan nama menu dan pendapatan tertinggi, serta menghitung total seluruh pendapatan.

Ringkasan Akhir: Menampilkan laporan singkat di konsol berupa total omset kantin, produk terlaris berdasarkan pendapatan, dan konfirmasi bahwa data sudah siap digunakan untuk visualisasi.

Alasan & Cara Mengatasi:

- Kami mengatasinya dengan berdiskusi kelompok untuk memahami konteks data:

 _Untuk kolom menu, nama produk adalah identitas utama. Transaksi tanpa nama menu tidak dapat diidentifikasi produknya, sehingga baris tersebut lebih tepat dihapus (dropna).

7. Menurutmu, mengapa keputusan membersihkan data (mis. menghapus vs mengisi data kosong) perlu didasarkan pada alasan yang jelas, bukan asal-asalan?

- Karena setiap aksi pembersihan data berdampak langsung pada akurasi dan keabsahan hasil analisis akhir (mencegah bias data).

 _Risiko Asal Menghapus (dropna sembarangan): Dapat menghilangkan informasi penting dan mengurangi ukuran dataset secara drastis, sehingga sampel data tidak lagi mewakili kondisi lapangan yang sebenarnya.

 _Risiko Asal Mengisi (fillna sembarangan): Mengisi data kosong dengan nilai yang tidak logis (misalnya mengisi jumlah terjual dengan angka rata-rata pada produk yang tidak laku) akan memanipulasi nilai statistik seperti mean, total omzet, dan estimasi keuntungan sehingga menghasilkan laporan yang menyesatkan.
